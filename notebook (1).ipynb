## 2. Theoretische Grundlagen

Im Folgenden definieren wir einige Grundbegriffe, um eine fundierte Basis für die Analyse zu schaffen.


### 2.1 Individuelle Förderung

Unter individueller Förderung versteht man laut Kultusministerkonferenz (KMK):

> *„Individuelle Förderung versucht, jedem Kind bestmögliche Lern- und Entwicklungschancen zu bieten. Dabei berücksichtigen Lehrkräfte die unterschiedlichen Lernvoraussetzungen, Interessen und Fähigkeiten der Schülerinnen und Schüler und unterstützen sie gezielt.“*

Dieser Anspruch bedeutet in der Praxis, dass Lehrkräfte den Unterricht so gestalten müssen, dass kein Kind über- oder unterfordert wird. Es geht nicht mehr darum, dass alle Schüler zum gleichen Zeitpunkt dasselbe Ziel auf demselben Weg erreichen. Stattdessen rücken differenzierte Lernangebote in den Fokus, die auf die spezifischen Stärken und Schwächen des Einzelnen eingehen.


### 2.2 Heterogenität

Unter Heterogenität versteht man im schulischen Kontext die Vielfalt der Lernvoraussetzungen innerhalb einer Gruppe. In unserer Analyse konzentrieren wir uns auf zwei wesentliche Ausprägungen, die den pädagogischen Alltag maßgeblich beeinflussen:

**Migrationshintergrund und sprachliche Diversität:**
Insbesondere seit dem Jahr 2016 ist die Schülerschaft im Hinblick auf ihre Herkunft und Muttersprache deutlich vielfältiger geworden. Für Lehrkräfte bedeutet dies eine zusätzliche Herausforderung, da sie neben den Fachinhalten oft auch grundlegende Sprachkenntnisse vermitteln müssen, um eine Teilhabe am Unterricht zu ermöglichen.

**Leistungsheterogenität:**
Die Schere zwischen den Lernniveaus innerhalb einer Klasse geht immer weiter auseinander. Während ein Teil der Schüler bereits komplexe Sachverhalte eigenständig löst, benötigt ein anderer Teil – selbst in höheren Schulformen – noch intensive Unterstützung bei Basiskompetenzen wie Lesen oder Schreiben.

Sowohl die sprachliche Vielfalt als auch die enormen Leistungsunterschiede führen dazu, dass die Anforderungen an die individuelle Förderung massiv gestiegen sind. Je unterschiedlicher die Startbedingungen der Kinder sind, desto zeitintensiver wird die notwendige Unterstützung für den Einzelnen.


### 2.3 PISA-Studie

Das *Programme for International Student Assessment* (PISA) ist eine international vergleichende Schulleistungsuntersuchung der OECD. Seit dem Jahr 2000 werden alle drei Jahre die Kompetenzen 15-jähriger Schülerinnen und Schüler in den drei Kernbereichen *Mathematik*, *Lesekompetenz* und *Naturwissenschaften* erhoben.

Die Ergebnisse werden auf einer Punkteskala dargestellt, deren Mittelwert ursprünglich auf 500 Punkte mit einer Standardabweichung von 100 Punkten kalibriert wurde. Eine Differenz von etwa 20 bis 25 Punkten entspricht dabei näherungsweise dem Lernfortschritt eines Schuljahres.

Für unsere Analyse ist PISA relevant, weil die Studie als etablierter Indikator für den tatsächlichen Lernerfolg dient. Veränderungen in den PISA-Werten zeigen, ob strukturelle Belastungen des Schulsystems – wie sinkende reale Lehrkapazität und wachsende Heterogenität – sich auch in messbaren Leistungseinbußen bei den Schülerinnen und Schülern niederschlagen.

## 3. Datenanalyse

Wir entwickeln die Argumentation entlang der Storyline. Alle Visualisierungen folgen einem einheitlichen Design (Farben, Typografie, Beschriftungen), damit sich die Aussagen klar einordnen lassen.


```python
"""
Setup: Imports, Daten-Loader und einheitliches Design-System.

Das Modul stellt drei Dinge bereit:
  1. Pfad-Konstanten für alle Excel-Quelldateien im Ordner Statistiken/.
  2. Eine einheitliche Farbpalette und matplotlib/plotly Styling-Defaults.
  3. Eine generische load()-Funktion für Excel-Dateien mit Standard-Aufbau
     (Header in Zeile 5, Spalte B = Schuljahr/Kategorie, Spalte C = Wert).
"""

from pathlib import Path

import matplotlib as mpl
import matplotlib.pyplot as plt
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio

# ---------------------------------------------------------------------------
# Datenquellen
# ---------------------------------------------------------------------------
DATA = Path("Statistiken")

F_LEHRER     = DATA / "Lehrer an allgemeinbildenden Schulen.xlsx"
F_SCHUELER   = DATA / "Schüler an allgemeinbildenden Schulen.xlsx"
F_BESCH      = DATA / "Anzahl der Lehrkräfte nach Beschäftigungsumfang und Schulart in Deutschland im Schuljahr 2023-2024.xlsx"
F_PROGNOSE   = DATA / "Prognose zum Lehrkräftemangel.xlsx"
F_PISA_M     = DATA / "Pisa - Mathematik.xlsx"
F_PISA_L     = DATA / "Pisa - Lesekompetenz.xlsx"
F_PISA_N     = DATA / "Pisa - Naturwissenschaften.xlsx"
F_AUSFALL    = DATA / "Umfrage unter Schulleitungen zur Zunahme von langfristigen Ausfällen aufgrund von Erkrankungen.xlsx"
F_HERAUSF    = DATA / "Umfrage Lehrkräfte zu Herausforderungen 2024 (1).xlsx"
F_INKLUSION  = DATA / "statistic_id1262397_inklusion_-schueler-mit-foerderbedarf-in-foerderschulen-und-regelschulen-bis-2023.xlsx"

# ---------------------------------------------------------------------------
# Design-System
# ---------------------------------------------------------------------------
COLOR = {
    "primary":  "#1f4e79",   # dunkles Blau – neutrale Hauptserie
    "accent":   "#c0392b",   # Rot – kritische Werte / Problemzustand
    "warning":  "#e67e22",   # Orange – Warnzone
    "positive": "#2e7d32",   # Grün – positive Werte
    "neutral":  "#7f8c8d",   # Grau – Sekundärinformation
    "muted":    "#bdc3c7",   # Hellgrau – Hintergrund/Raster
    "ink":      "#2c3e50",   # Textfarbe
    "bg":       "#ffffff",
}

FONT_FAMILY = "DejaVu Sans, Helvetica, Arial, sans-serif"

# matplotlib-Defaults
mpl.rcParams.update({
    "figure.facecolor": COLOR["bg"],
    "axes.facecolor":   COLOR["bg"],
    "axes.edgecolor":   COLOR["ink"],
    "axes.labelcolor":  COLOR["ink"],
    "axes.titlecolor":  COLOR["ink"],
    "axes.titleweight": "bold",
    "axes.titlesize":   14,
    "axes.labelsize":   11,
    "xtick.color":      COLOR["ink"],
    "ytick.color":      COLOR["ink"],
    "axes.spines.top":   False,
    "axes.spines.right": False,
    "grid.color":       COLOR["muted"],
    "grid.linestyle":   "-",
    "grid.linewidth":   0.6,
    "grid.alpha":       0.5,
    "font.family":      "sans-serif",
    "font.size":        11,
})

# plotly-Defaults: einheitliches Template
pio.templates["projekt"] = go.layout.Template(
    layout=dict(
        font=dict(family=FONT_FAMILY, size=13, color=COLOR["ink"]),
        title=dict(font=dict(size=18, color=COLOR["ink"]), x=0.02, xanchor="left"),
        plot_bgcolor=COLOR["bg"],
        paper_bgcolor=COLOR["bg"],
        colorway=[COLOR["primary"], COLOR["accent"], COLOR["positive"],
                  COLOR["warning"], COLOR["neutral"]],
        xaxis=dict(showline=True, linecolor=COLOR["ink"], linewidth=1,
                   gridcolor=COLOR["muted"], zeroline=False, ticks="outside"),
        yaxis=dict(showline=True, linecolor=COLOR["ink"], linewidth=1,
                   gridcolor=COLOR["muted"], zeroline=False, ticks="outside"),
        legend=dict(bgcolor="rgba(255,255,255,0.85)", bordercolor=COLOR["muted"], borderwidth=0),
        margin=dict(l=70, r=40, t=80, b=70),
    )
)
pio.templates.default = "plotly_white+projekt"


# ---------------------------------------------------------------------------
# Daten-Loader
# ---------------------------------------------------------------------------
def load(path, *, value_col_name="Anzahl"):
    """Liest eine Excel-Datei mit Standard-Aufbau aus Statistiken/.

    Erwartet im Sheet 'Daten':
      - 5 Meta-/Headerzeilen (skiprows=5),
      - Spalte B = Schuljahr / Kategorie,
      - Spalte C = numerischer Wert.

    Liefert ein DataFrame mit zwei Spalten: ['Schuljahr', value_col_name].
    Leerzeilen am Tabellenende werden entfernt.
    """
    df = (pd.read_excel(path, sheet_name="Daten",
                        header=None, usecols=[1, 2], skiprows=5)
            .dropna())
    df.columns = ["Schuljahr", value_col_name]
    return df.reset_index(drop=True)
```

### 3.1 Erste Beobachtung – Die Schüler-Lehrer-Relation sinkt

Auf den ersten Blick deutet die Statistik auf eine *positive* Entwicklung hin: Über die letzten 15 Jahre ist die durchschnittliche Schüler-Lehrer-Relation kontinuierlich gesunken. Rein rechnerisch betreut eine Lehrkraft heute weniger Kinder als noch vor einer Dekade.


```python
# Schüler-Lehrer-Relation (SLR) – Slope-Chart: erstes vs. letztes Schuljahr.
#
# Datenbasis: amtliche Statistik zu Schüler- und Lehrkräftezahlen an
# allgemeinbildenden Schulen. Aus beiden Reihen wird per Inner-Join nur der
# Zeitraum behalten, für den beide Werte vorliegen. Die SLR ergibt sich
# rechnerisch als Schüler / Lehrkräfte (ohne Korrektur für den
# Beschäftigungsumfang – das diskutieren wir im nächsten Abschnitt).
#
# Die Darstellung als Slope-Chart reduziert die Botschaft bewusst auf zwei
# Punkte: Start- und Endwert des betrachteten Zeitraums. Die Veränderung wird
# als Δ in der Mitte hervorgehoben.

lehrer   = load(F_LEHRER,   value_col_name="Lehrkraefte").astype({"Lehrkraefte": int})
schueler = load(F_SCHUELER, value_col_name="Schueler").astype({"Schueler": int})

df_slr = (schueler.merge(lehrer, on="Schuljahr", how="inner")
                  .sort_values("Schuljahr")
                  .reset_index(drop=True))
df_slr["SchuelerProLehrer"] = df_slr["Schueler"] / df_slr["Lehrkraefte"]

# ---------------------------------------------------------------------------
# Visualisierung: Slope-Chart erstes Jahr -> letztes Jahr.
# ---------------------------------------------------------------------------
first, last = df_slr.iloc[0], df_slr.iloc[-1]
x = [0, 1]
y = [first["SchuelerProLehrer"], last["SchuelerProLehrer"]]

fig, ax = plt.subplots(figsize=(7, 6))

ax.plot(x, y, color=COLOR["primary"], linewidth=3,
        marker="o", markersize=14, zorder=3)

# Wertelabels neben den Endpunkten
ax.text(-0.05, y[0], f"{y[0]:.1f}", ha="right", va="center",
        fontsize=22, fontweight="bold", color=COLOR["primary"])
ax.text(1.05, y[1], f"{y[1]:.1f}", ha="left", va="center",
        fontsize=22, fontweight="bold", color=COLOR["primary"])

# Delta-Pille in Linienmitte
ax.text(0.5, (y[0] + y[1]) / 2 + 0.05, f"Δ {y[1] - y[0]:+.1f}",
        ha="center", va="bottom",
        fontsize=14, fontweight="bold", color="white",
        bbox=dict(boxstyle="round,pad=0.35",
                  facecolor=COLOR["primary"], edgecolor="none"))

ax.set_xticks(x)
ax.set_xticklabels([first["Schuljahr"], last["Schuljahr"]],
                   fontsize=12, fontweight="bold")
ax.set_xlim(-0.4, 1.4)
ax.set_ylim(min(y) - 0.8, max(y) + 0.8)
ax.set_ylabel("Schüler je Lehrkraft")
ax.grid(axis="y")
ax.set_axisbelow(True)
for spine in ("bottom",):
    ax.spines[spine].set_visible(False)
ax.tick_params(axis="x", length=0)

ax.set_title("Schüler je Lehrkraft – Anfang vs. Ende\n"
             f"{first['Schuljahr']}  →  {last['Schuljahr']}",
             loc="left", pad=14)

fig.text(0.01, -0.02,
         "Datenquelle: Statistisches Bundesamt – allgemeinbildende Schulen.",
         fontsize=9, color=COLOR["neutral"])

plt.tight_layout()
plt.show()
```

**Befund:** Die SLR ist seit 2010/11 um rund 1 Schüler pro Lehrkraft gesunken. Wenn weniger Kinder auf eine Lehrkraft kommen, sollte mehr Zeit für individuelle Förderung übrig bleiben – so die naive Interpretation. Im nächsten Schritt prüfen wir, ob die Kennzahl die tatsächliche Lehrkapazität widerspiegelt.


**Visualisierungswahl – Slope-Chart:** Die eigentliche Pointe steckt im *absoluten Wert*: Eine SLR von rund 10,5 bis 11,5 Schülerinnen und Schülern pro Lehrkraft liegt deutlich unter dem, was man intuitiv erwarten würde – die meisten Menschen denken hier eher an Klassengrößen von 25 bis 30. Der Slope-Chart stellt diese zwei Zahlen groß und prominent in den Vordergrund, sodass die überraschend niedrige Größenordnung sofort ins Auge fällt. Dass der Wert über die Jahre zusätzlich gesunken ist, verstärkt diesen Eindruck nur noch: Auf den ersten Blick wirkt das System komfortabel ausgestattet.

**Warum nicht anders?** Ein klassisches Liniendiagramm über alle Jahre haben wir verworfen, weil die jährlichen Schwankungen vom eigentlichen Punkt – der überraschend niedrigen Größenordnung und ihrer Bewegungsrichtung – ablenken würden. Ein Balkendiagramm würde zwar einen Vorher-Nachher-Vergleich erlauben, transportiert aber die *Richtung* der Veränderung schlechter als eine schräg verlaufende Linie und drückt die Zahlen optisch kleiner. Eine reine Tabelle wäre exakt, aber visuell schwach – die beiden zentralen Werte sollen *wirken*, nicht nur dastehen.

**Nachteile, die wir in Kauf nehmen:** Der Slope-Chart unterschlägt den Verlauf zwischen den Endpunkten und suggeriert implizit Linearität. Außerdem hängt die Aussage von der Wahl der Randjahre ab – ein längeres oder kürzeres Fenster könnte den Trend anders aussehen lassen. Diese Reduktion ist hier akzeptabel, weil die folgenden Abschnitte den Befund ohnehin gezielt relativieren: Wir wollen den naiven, scheinbar positiven Eindruck zunächst klar herausstellen, um ihn anschließend bewusst aufzubrechen.

**Technischer Aufbau:** Die Daten werden mit *pandas* eingelesen (über den projektweiten `load()`-Helper auf Basis von `pd.read_excel`), per Inner-Join auf gemeinsame Schuljahre reduziert und die SLR als Quotient berechnet. Die Visualisierung selbst nutzt *matplotlib* (`pyplot`): `ax.plot` für die Slope-Linie mit Markern, `ax.text` für die großen Wertelabels an den Endpunkten sowie eine in `bbox` eingebettete Pille für das Δ-Label. Achsen, Spines und Ticks werden über das im Setup definierte einheitliche `rcParams`-Styling konsistent zum Rest des Berichts gestaltet. Bewusst kein *plotly* an dieser Stelle: Der Slope-Chart braucht keine Interaktivität – eine statische, drucktaugliche Darstellung passt besser zur reduzierten Botschaft.


### 3.2 Relativierung – Beschäftigungsumfang der Lehrkräfte

Die offizielle SLR behandelt jede Lehrkraft als gleichwertige Einheit – unabhängig davon, ob sie Vollzeit, Teilzeit oder stundenweise unterrichtet. Tatsächlich arbeitet nur rund jede zweite Lehrkraft in Vollzeit. Damit überschätzt die rechnerische Kennzahl die real verfügbare Lehrkapazität spürbar.

Die folgende Waffle-Visualisierung zeigt, wie sich 100 Lehrkräfte einer Schulart auf die drei Beschäftigungsformen aufteilen. Jedes Quadrat steht für eine Lehrkraft. Über das Dropdown lässt sich die Schulart wechseln.


```python
# Beschäftigungsumfang als Waffle-Chart (10 x 10 Quadrate = 100 Lehrkräfte).
#
# Datenbasis: Statistisches Bundesamt, Schuljahr 2023/24, Lehrkräfte nach
# Beschäftigungsumfang und Schulart. Pro Schulart werden die drei Kategorien
# auf 100 Quadrate gerundet; Rundungsdifferenzen landen in der dritten
# Kategorie ("Stundenweise"), damit die Summe exakt 100 ergibt.

besch = (pd.read_excel(F_BESCH, sheet_name="Daten",
                       usecols=[1, 2, 3, 4], skiprows=4)
           .rename(columns={"Unnamed: 1": "Schulart"})
           .dropna(subset=["Schulart"]))
besch["Schulart"] = besch["Schulart"].replace({"Gynmnasien": "Gymnasien"})

CATS_FULL = ["Vollzeitbeschäftigte", "Teilzeitbeschäftigte", "Stundenweise Beschäftigte"]
CATS = ["Vollzeit", "Teilzeit", "Stundenweise"]
COL_WAFFLE = {"Vollzeit":      COLOR["primary"],
              "Teilzeit":      COLOR["warning"],
              "Stundenweise":  COLOR["accent"]}
besch["Total"] = besch[CATS_FULL].sum(axis=1)


def waffle_counts(row):
    """Rundet die Anteile auf 100 Quadrate (Rest -> Stundenweise)."""
    vz  = round(row["Vollzeitbeschäftigte"]    / row["Total"] * 100)
    tz  = round(row["Teilzeitbeschäftigte"]    / row["Total"] * 100)
    std = 100 - vz - tz
    return vz, tz, std


def waffle_xy(vz, tz, std):
    """Erzeugt die x/y-Positionen der Quadrate – zeilenweise von oben nach unten."""
    sequence = ["Vollzeit"] * vz + ["Teilzeit"] * tz + ["Stundenweise"] * std
    xs = {c: [] for c in CATS}
    ys = {c: [] for c in CATS}
    for i, c in enumerate(sequence):
        xs[c].append(i % 10)
        ys[c].append(9 - i // 10)
    return xs, ys


# Initialdarstellung: alle Schularten zusammen
init_row = besch[besch["Schulart"] == "Lehrkräfte insgesamt"].iloc[0]
vz0, tz0, std0 = waffle_counts(init_row)
xs0, ys0 = waffle_xy(vz0, tz0, std0)
init_counts = {"Vollzeit": vz0, "Teilzeit": tz0, "Stundenweise": std0}

fig = go.Figure()
for cat in CATS:
    fig.add_trace(go.Scatter(
        x=xs0[cat], y=ys0[cat], mode="markers",
        name=f"{cat} ({init_counts[cat]})",
        marker=dict(size=34, color=COL_WAFFLE[cat], symbol="square",
                    line=dict(color="white", width=2)),
        hovertemplate=f"<b>{cat}</b><extra></extra>",
    ))

# Dropdown-Buttons: pro Schulart wird die Datenreihe neu gesetzt
buttons = []
for sa in besch["Schulart"]:
    r = besch[besch["Schulart"] == sa].iloc[0]
    vz, tz, std = waffle_counts(r)
    xs, ys = waffle_xy(vz, tz, std)
    buttons.append(dict(
        label=sa, method="update",
        args=[
            {"x":    [xs["Vollzeit"], xs["Teilzeit"], xs["Stundenweise"]],
             "y":    [ys["Vollzeit"], ys["Teilzeit"], ys["Stundenweise"]],
             "name": [f"Vollzeit ({vz})", f"Teilzeit ({tz})", f"Stundenweise ({std})"]},
            {"title.text":
                f"<b>{sa}</b> – Beschäftigungsumfang je 100 Lehrkräfte (Schuljahr 2023/24)"}
        ]))

fig.update_layout(
    title=dict(text="<b>Lehrkräfte insgesamt</b> – Beschäftigungsumfang je 100 Lehrkräfte (Schuljahr 2023/24)"),
    xaxis=dict(visible=False, range=[-0.7, 9.7]),
    yaxis=dict(visible=False, range=[-0.7, 9.7], scaleanchor="x"),
    height=620, width=780,
    legend=dict(orientation="h", yanchor="bottom", y=-0.05,
                xanchor="center", x=0.5, font=dict(size=12)),
    updatemenus=[dict(buttons=buttons, x=1.0, xanchor="right",
                      y=1.12, yanchor="top", direction="down", showactive=True)],
    margin=dict(l=20, r=20, t=90, b=70),
    annotations=[dict(text="Datenquelle: Statistisches Bundesamt, Schuljahr 2023/24",
                      x=0, y=-0.13, xref="paper", yref="paper",
                      showarrow=False, font=dict(size=11, color=COLOR["neutral"]),
                      xanchor="left")],
)
fig.show()
```

**Befund:** Über alle Schularten hinweg arbeiten nur rund **52 von 100 Lehrkräften in Vollzeit**. Rechnet man die Teilzeit- und Stundenweise-Anteile in *Vollzeitäquivalente* um, schrumpft die reale Lehrkapazität deutlich gegenüber dem Kopfzahl-Eindruck der SLR.


**Visualisierungswahl – Waffle-Chart:** Die Aussage „nur rund jede zweite Lehrkraft arbeitet Vollzeit“ soll *greifbar* werden. In einem 10 × 10-Raster steht jedes Quadrat für eine Lehrkraft – die Leserin kann die Anteile abzählen statt sie zu schätzen. Damit wird aus einer abstrakten Prozentangabe ein konkretes Bild von 100 Personen.

**Warum nicht anders?** Ein Tortendiagramm haben wir bewusst vermieden: Winkelflächen sind notorisch schlecht zu vergleichen, gerade bei drei Kategorien mit ähnlichen Anteilen. Ein gestapelter Balken wäre kompakter, transportiert die Botschaft aber abstrakter und verliert den Bezug zur Einzelperson. Mehrere Balken nebeneinander für alle Schularten gleichzeitig hätten die Darstellung überladen und den Effekt pro Schulart verwischt. Das Dropdown löst dieses Problem: Eine Schulart steht im Fokus, der Wechsel ist trotzdem nur einen Klick entfernt.

**Nachteile, die wir in Kauf nehmen:** Die Rundung auf 100 Quadrate erzeugt minimale Verzerrungen, die wir bewusst in die Kategorie *Stundenweise* lenken, damit die Summe exakt aufgeht. Außerdem setzt die Visualisierung Interaktivität voraus – in einer gedruckten Fassung wäre nur die initial gewählte Schulart sichtbar. Schließlich erfordert der direkte Vergleich mehrerer Schularten sequentielles Umschalten statt eines simultanen Side-by-Side-Vergleichs. Für unsere Storyline überwiegt der Vorteil der greifbaren Darstellung diese Einschränkungen.

**Technischer Aufbau:** Die Daten werden mit *pandas* direkt aus der Excel-Quelle eingelesen (`pd.read_excel` mit gezielter Spaltenauswahl, Bereinigung über `dropna` und Korrektur eines Tippfehlers in den Schulart-Bezeichnungen). Die Visualisierung baut auf *plotly* (`graph_objects`): Statt eines dedizierten Waffle-Moduls nutzen wir `go.Scatter` mit `marker.symbol='square'` und einer großen Markergröße – jedes Symbol wird über berechnete x/y-Koordinaten exakt auf einem 10 × 10-Raster platziert. Hilfsfunktionen `waffle_counts` und `waffle_xy` kapseln die Rundung auf 100 Quadrate und die Positionsberechnung. Der Schulartwechsel ist als `updatemenus`-Dropdown realisiert, das per `method='update'` die Datenreihen und den Titel austauscht, ohne die Figur neu zu rendern. Hier ist *plotly* bewusst gewählt: Die Interaktivität (Hover, Schulart-Dropdown) ist Teil des Storytellings – statisch wäre die Botschaft nur halb so stark.


### 3.3 Zusätzliche Belastung – Wachsender Förderbedarf an allgemeinen Schulen

Während die reale Lehrkapazität geringer ausfällt als die SLR suggeriert, sind gleichzeitig die Anforderungen gestiegen. Ein zentraler Treiber: Schülerinnen und Schüler mit sonderpädagogischem Förderbedarf werden zunehmend an allgemeinen Schulen statt an Förderschulen unterrichtet (Inklusion). Die Gesamtzahl der Kinder mit Förderbedarf wächst, der Anteil an Förderschulen bleibt dabei nahezu konstant – der gesamte Zuwachs landet in den Regelklassen.

Die folgende Darstellung zeigt diese Entwicklung als Liniendiagramm. Die vertikale Differenz zwischen den beiden Linien entspricht der Anzahl an Kindern mit Förderbedarf, die an allgemeinen Schulen unterrichtet werden.


```python
# Inklusion: Schüler mit sonderpädagogischem Förderbedarf 2013-2023, aufgeteilt
# nach Förderschule und allgemeiner Schule.
#
# Datenbasis: Statistik "Inklusion - Schüler mit Förderbedarf in Förderschulen
# und Regelschulen bis 2023". Aus Spalte B/C/D werden Jahr, Förderschule und
# Allgemeine Schule eingelesen; die Gesamtzahl ist deren Summe.
#
# Visualisierung: Liniendiagramm mit zwei Reihen (Gesamt, Förderschule). Die
# vertikale Differenz zwischen beiden Linien entspricht den an allgemeinen
# Schulen unterrichteten Förderbedarfen und wird über farbig gefüllte Flächen
# sowie Doppelpfeile an den Randjahren explizit ausgewiesen.

inklusion = (pd.read_excel(F_INKLUSION, sheet_name="Daten",
                           header=None, usecols=[1, 2, 3], skiprows=5)
               .dropna())
inklusion.columns = ["Jahr", "Foerderschule", "AllgemeineSchule"]
inklusion = inklusion.astype({"Jahr": int, "Foerderschule": int, "AllgemeineSchule": int})
inklusion["Gesamt"] = inklusion["Foerderschule"] + inklusion["AllgemeineSchule"]
inklusion = inklusion.sort_values("Jahr").reset_index(drop=True)

j_start, j_end = int(inklusion["Jahr"].min()), int(inklusion["Jahr"].max())
row_start = inklusion.loc[inklusion["Jahr"] == j_start].iloc[0]
row_end   = inklusion.loc[inklusion["Jahr"] == j_end].iloc[0]


def de(n):
    """Tausendertrennzeichen im deutschen Format."""
    return f"{int(n):,}".replace(",", ".")


fig, ax = plt.subplots(figsize=(13, 8))

# Hauptlinien
ax.plot(inklusion["Jahr"], inklusion["Gesamt"],
        marker="o", color=COLOR["primary"], linewidth=3,
        markersize=8, label="Gesamt (Förderbedarf)")
ax.plot(inklusion["Jahr"], inklusion["Foerderschule"],
        marker="s", color=COLOR["ink"], linewidth=3,
        markersize=8, label="Förderschule")

# Flächen
ax.fill_between(inklusion["Jahr"], 0, inklusion["Foerderschule"],
                color=COLOR["positive"], alpha=0.18,
                label="An Förderschulen")
ax.fill_between(inklusion["Jahr"], inklusion["Foerderschule"], inklusion["Gesamt"],
                color=COLOR["accent"], alpha=0.22,
                label="An allgemeinen Schulen")

# Endpunkt-Marker (akzentuiert)
ax.scatter([j_start, j_end], [row_start["Gesamt"], row_end["Gesamt"]],
           color=COLOR["primary"], zorder=5, s=130, edgecolor="white", linewidth=2)
ax.scatter([j_start, j_end], [row_start["Foerderschule"], row_end["Foerderschule"]],
           color=COLOR["ink"], zorder=5, s=130, edgecolor="white", linewidth=2)

# Wertelabels (Gesamt oben, Förderschule unten)
label_offset = 17_000
ax.text(j_start - 0.2, row_start["Gesamt"] + label_offset, de(row_start["Gesamt"]),
        color=COLOR["primary"], fontsize=16, fontweight="bold", ha="right", va="center")
ax.text(j_end + 0.2, row_end["Gesamt"] + label_offset, de(row_end["Gesamt"]),
        color=COLOR["primary"], fontsize=16, fontweight="bold", ha="left", va="center")
ax.text(j_start - 0.2, row_start["Foerderschule"] - label_offset, de(row_start["Foerderschule"]),
        color=COLOR["ink"], fontsize=16, fontweight="bold", ha="right", va="center")
ax.text(j_end + 0.2, row_end["Foerderschule"] - label_offset, de(row_end["Foerderschule"]),
        color=COLOR["ink"], fontsize=16, fontweight="bold", ha="left", va="center")

# Doppelpfeile + Boxen für die Differenz an den Randjahren
for jahr, row, ha, dx in [
    (j_start, row_start, "left",  +0.25),
    (j_end,   row_end,   "right", -0.25),
]:
    ax.annotate("", xy=(jahr, row["Gesamt"] - 4_000),
                xytext=(jahr, row["Foerderschule"] + 4_000),
                arrowprops=dict(arrowstyle="<->", color=COLOR["accent"], lw=2.2))
    diff = int(row["Gesamt"] - row["Foerderschule"])
    ax.text(jahr + dx, (row["Gesamt"] + row["Foerderschule"]) / 2,
            f"An allg. Schulen:\n{de(diff)}",
            color=COLOR["accent"], fontsize=11, fontweight="bold",
            ha=ha, va="center",
            bbox=dict(boxstyle="round,pad=0.35", fc="white",
                      ec=COLOR["accent"], lw=1.4))

# Achsen, Limits, Beschriftung
ax.set_xticks(inklusion["Jahr"])
ax.set_xlim(j_start - 1.2, j_end + 1.2)
ax.set_ylim(0, inklusion["Gesamt"].max() * 1.18)
ax.set_xlabel("Jahr")
ax.set_ylabel("Anzahl Schüler")
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: de(x)))
ax.grid(axis="y")
ax.set_axisbelow(True)
ax.legend(loc="lower left", frameon=False, fontsize=10)

ax.set_title(f"Inklusion: Förderbedarf an Förder- vs. allgemeinen Schulen, {j_start}-{j_end}\n"
             "Vertikale Differenz = an allgemeinen Schulen unterrichtete Kinder",
             loc="left", pad=14)

fig.text(0.01, -0.02,
         "Datenquelle: Sekretariat der Kultusministerkonferenz (KMK) / Statista, "
         "Sonderpädagogische Förderung nach Schulart.",
         fontsize=9, color=COLOR["neutral"])

plt.tight_layout()
plt.show()
```

**Befund:** Die Gesamtzahl der Schüler mit sonderpädagogischem Förderbedarf ist von **500.544** (2013) auf **608.097** (2023) gestiegen. Die Zahl an Förderschulen blieb dabei nahezu unverändert (**343.343** → **344.363**), der gesamte Zuwachs entfiel auf die allgemeinen Schulen: dort hat sich der Förderbedarf von **157.201** auf **263.734** Kinder fast verdoppelt. Lehrkräfte an Regelschulen begegnen damit deutlich häufiger Kindern mit besonderem Förderbedarf – ohne dass viele von ihnen sonderpädagogisch ausgebildet wären.


**Visualisierungswahl – Liniendiagramm mit Flächenfärbung:** Die Darstellung bildet die stetige Entwicklung über das gesamte Jahrzehnt ab und macht die Dynamik der beiden Reihen auf einen Blick sichtbar: Während die Förderschul-Linie nahezu stagniert, steigt die Gesamtzahl kontinuierlich an. Die Flächenfärbung zwischen den Linien visualisiert die Anteile direkt – das Auge erfasst sofort, wie groß der rote Anteil für die allgemeinen Schulen im Verlauf geworden ist. Die explizit eingezeichneten Doppelpfeile an den Randjahren beziffern die Differenz zusätzlich exakt.

**Warum nicht anders?** Ein gestapeltes Balkendiagramm hätte die Anteile pro Jahr ebenfalls gezeigt, aber den kontinuierlichen Verlauf in diskrete Säulen zerlegt – die Botschaft „kontinuierliches Wachstum bei stagnierender Förderschule“ wäre verloren gegangen. Zwei separate Linien ohne Flächenfärbung würden den Vergleich zwischen den Reihen erschweren, weil das Auge die vertikale Differenz nur schätzen kann. Ein Slope-Chart wie in 3.1 würde die Zwischenjahre unterschlagen und die Frage offen lassen, ob der Anstieg gleichmäßig oder sprunghaft verläuft.

**Nachteile, die wir in Kauf nehmen:** Durch die zwei Linien und die dazwischenliegenden Flächen wirkt die Grafik auf den ersten Blick komplexer als ein einzelnes Standarddiagramm. Die exakten Zahlenwerte zwischen den Randjahren lassen sich nicht direkt ablesen, sondern müssen über das Achsenraster geschätzt werden. Für die Kernbotschaft – Trend und Größenordnung der Verschiebung – ist diese Detailgenauigkeit aber nicht entscheidend.

**Technischer Aufbau:** Die Daten werden mit *pandas* eingelesen, unvollständige Zeilen via `dropna()` entfernt und Jahres- sowie Schülerzahlspalten mit `astype(int)` typisiert. Die Gesamtzahl entsteht als Summenspalte aus Förderschule und allgemeiner Schule. Die Visualisierung nutzt *matplotlib*: `ax.plot` zeichnet die beiden Linienreihen mit unterschiedlichen Markern (`"o"` und `"s"`). Der entscheidende visuelle Kniff liegt in zwei `ax.fill_between`-Aufrufen – einer füllt die Fläche bis zur Förderschul-Linie grün, der zweite die Spanne darüber bis zur Gesamtlinie rot. An den Randjahren markiert eine Schleife über `ax.annotate(arrowstyle="<->")` die vertikalen Doppelpfeile, während `ax.text` mit `bbox`-Argument die exakten Differenzwerte in formatierten Boxen platziert. Eine Hilfsfunktion `de(n)` formatiert alle Zahlen mit deutschem Tausendertrennzeichen, `plt.FuncFormatter` wendet sie auf die Y-Achse an. Bewusst kein *plotly*: Der Fokus liegt auf einer klaren Trend-Aussage – Interaktivität wäre Ablenkung.


### 3.4 Quantifizierung – Prognose zum Lehrkräftemangel bis 2035

Wie groß ist die Lücke zwischen Bedarf und Angebot tatsächlich? Die KMK-Modellrechnung zeigt: 2024 fehlen rund **17.400 Lehrkräftestellen**. Die Lücke schließt sich laut Prognose bis Anfang der 2030er Jahre – allerdings unter der Annahme, dass Studierendenzahlen, Pensionierungen und Zuwanderung im modellierten Korridor bleiben.


```python
# Prognose: Differenz von Lehrkräfteangebot und -bedarf 2024-2035.
#
# Negative Werte = Lehrkräftemangel, positive Werte = Überschuss.
# Farbverlauf signalisiert den Schweregrad:
#   rot       <= -10.000   (akuter Mangel)
#   orange    <=  -5.000   (deutlicher Mangel)
#   gelb      <       0    (leichter Mangel)
#   grün      >=      0    (Überschuss)

prognose = (load(F_PROGNOSE).astype({"Schuljahr": int, "Anzahl": int})
                            .sort_values("Schuljahr")
                            .reset_index(drop=True))

jahre = prognose["Schuljahr"].tolist()
werte = prognose["Anzahl"].tolist()

def color_for(v):
    if v <= -10_000: return COLOR["accent"]
    if v <=  -5_000: return COLOR["warning"]
    if v <       0:  return "#f1c40f"
    return COLOR["positive"]

farben = [color_for(v) for v in werte]

fig = go.Figure()
fig.add_trace(go.Bar(
    x=jahre, y=werte,
    text=[f"{v:+,}".replace(",", ".") for v in werte],
    textposition="outside",
    marker_color=farben,
    marker_line_color=COLOR["ink"], marker_line_width=0.8,
    hovertemplate="<b>Jahr %{x}</b><br>%{y:+,} Stellen<extra></extra>",
))

y_min = (min(werte) // 1000) * 1000 - 3000
y_max = (max(werte) // 1000 + 1) * 1000 + 3000

fig.update_layout(
    title=dict(text=f"<b>Prognose zum Lehrkräftemangel bis {max(jahre)}</b><br>"
                     "<span style='font-size:13px;color:#7f8c8d'>"
                     "Differenz aus Lehrkräfteangebot und Lehrkräftebedarf</span>"),
    width=1100, height=620,
    showlegend=False,
    annotations=[dict(text="Datenquelle: Kultusministerkonferenz (KMK), Modellrechnung 2024.",
                      x=0, y=-0.18, xref="paper", yref="paper",
                      showarrow=False, font=dict(size=11, color=COLOR["neutral"]),
                      xanchor="left")],
)
fig.update_xaxes(dtick=1, title="Jahr")
fig.update_yaxes(title="Differenz Angebot – Bedarf (Stellen)",
                 zeroline=True, zerolinecolor=COLOR["ink"], zerolinewidth=1.5,
                 range=[y_min, y_max])
fig.show()
```

**Befund:** Die Modellrechnung erwartet **bis 2030 jedes Jahr einen Mangel** an Lehrkräften. Erst ab 2033 sind nach dieser Prognose leichte Überschüsse zu erwarten. Bis dahin bleibt individuelle Förderung systemisch unter Druck.


**Visualisierungswahl – Balkendiagramm mit Schwellwert-Farbcodierung:** Jedes Jahr spielt eine eigene Rolle in der Prognose, die einzelnen Werte sollen direkt ablesbar sein. Die vertikale Ausrichtung um eine Nulllinie bildet die Schere zwischen Angebot und Bedarf unmittelbar ab – negative Balken nach unten, positive nach oben. Die exakten Zahlen stehen direkt am Balkenende, sodass der Leser nicht schätzen muss. Der Übergang von langen negativen Balken zu den ersten positiven Werten 2033 markiert die erhoffte Wende auf dem Arbeitsmarkt klar und sichtbar. Das farbliche Ampelsystem (rot/orange/gelb/grün) codiert den Schweregrad des Mangels intuitiv.

**Warum nicht anders?** Eine durchgehende Linie hätte zwar den kontinuierlichen Trend dynamischer dargestellt, aber die einzelnen Jahreswerte unleserlich gemacht – bei einer Prognose, in der einzelne Wendejahre (2030, 2033) die Aussage tragen, ist das ein Nachteil. Zwei nebeneinanderliegende Reihen für Angebot und Bedarf hätten die Differenz zwar abgebildet, aber den Leser zur Subtraktion gezwungen. Eine Tabelle wäre exakt, aber blind für den Trend. Die Schwellwert-Farben statt einer einheitlichen Balkenfarbe sind bewusst gewählt: Sie machen die Schweregrade des Mangels auf einen Blick erfassbar.

**Nachteile, die wir in Kauf nehmen:** Die freistehenden Einzelbalken rücken den fließenden Charakter des Trends in den Hintergrund. Außerdem fokussiert die Grafik nur auf die Differenz – die absoluten Größen von Angebot und Bedarf bleiben verborgen, was eine relative Einordnung („17.400 von wie vielen?“) erschwert. Die diskreten Farbschwellen (5.000/10.000) sind ein gewählter Schnitt; ein gradueller Farbverlauf wäre objektiver, aber visuell schwerer zu lesen.

**Technischer Aufbau:** Die Daten werden über den `load()`-Helper aus dem Setup eingelesen, nach `Schuljahr` sortiert und in Listen für `x`/`y` überführt. Der visuelle Kern liegt in der Hilfsfunktion `color_for(v)`, die je nach Wert eine Schwellwert-Farbe aus der Projekt-Palette zuweist; das Ergebnis ist eine Liste `farben`, die per `marker_color` an die Balkenreihe übergeben wird. Die Visualisierung selbst nutzt *plotly* (`graph_objects`): `go.Bar` zeichnet die Balken, der Parameter `text` mit Listenabstraktion erzeugt die Wertelabels, `textposition="outside"` platziert sie sauber außerhalb. `fig.update_yaxes(zeroline=True)` hebt die kritische Nulllinie hervor. Das `hovertemplate` sorgt für eine sprechende Tooltip-Darstellung beim Überfahren der Balken. *plotly* hier bewusst gewählt: Hover-Werte sind nützlich für ein Diagramm mit zwölf Datenpunkten, in denen sich der Leser auf einzelne Jahre konzentrieren möchte.


### 3.5 Auswirkungen auf die Schüler – PISA-Ergebnisse seit 2009

Wenn Lehrkapazität sinkt und Heterogenität steigt, sollten messbare Auswirkungen auf den Lernerfolg sichtbar werden. Die PISA-Studie misst alle drei Jahre die Kernkompetenzen 15-jähriger Schülerinnen und Schüler in Mathematik, Lesekompetenz und Naturwissenschaften. Wir betrachten den Zeitraum seit 2009.


```python
# PISA-Punktwerte für Deutschland seit 2009 in den drei Hauptdomänen.
#
# Pro Domäne wird eine Excel-Datei eingelesen; gemeinsame Jahre werden per
# Schnittmenge bestimmt und ab 2009 dargestellt.

pisa_m = load(F_PISA_M).astype({"Schuljahr": int, "Anzahl": int})
pisa_l = load(F_PISA_L).astype({"Schuljahr": int, "Anzahl": int})
pisa_n = load(F_PISA_N).astype({"Schuljahr": int, "Anzahl": int})

jahre = sorted(set(pisa_m["Schuljahr"]) & set(pisa_l["Schuljahr"]) & set(pisa_n["Schuljahr"]))
jahre = [y for y in jahre if y >= 2009]


def vals_for(df):
    return [int(df.loc[df["Schuljahr"] == y, "Anzahl"].iloc[0]) for y in jahre]


pisa = {
    "Mathematik":          (vals_for(pisa_m), COLOR["primary"]),
    "Lesekompetenz":       (vals_for(pisa_l), COLOR["accent"]),
    "Naturwissenschaften": (vals_for(pisa_n), COLOR["positive"]),
}

fig = go.Figure()
for fach, (werte, farbe) in pisa.items():
    fig.add_trace(go.Scatter(
        x=jahre, y=werte, mode="lines+markers",
        name=fach,
        line=dict(color=farbe, width=3),
        marker=dict(size=10, line=dict(color="white", width=1.5)),
        hovertemplate=f"<b>{fach}</b><br>Jahr: %{{x}}<br>Punktzahl: %{{y}}<extra></extra>",
    ))

# Annotation: Veränderung 2009 -> 2022 je Domäne
ende = jahre[-1]; anfang = jahre[0]
deltas = []
for fach, (werte, farbe) in pisa.items():
    delta = werte[-1] - werte[0]
    deltas.append(f"<span style='color:{farbe}'><b>{fach}</b>: {delta:+d} Punkte</span>")
subtitle = " &nbsp; · &nbsp; ".join(deltas)

alle_werte = [v for werte, _ in pisa.values() for v in werte]
y_min = (min(alle_werte) // 10) * 10 - 10
y_max = (max(alle_werte) // 10 + 1) * 10 + 10

fig.update_layout(
    title=dict(text=f"<b>PISA-Kernkompetenzen Deutschland, {anfang} – {ende}</b><br>"
                    f"<span style='font-size:13px'>Δ {anfang} → {ende}: {subtitle}</span>"),
    width=1100, height=600,
    hovermode="x unified",
    legend=dict(orientation="h", yanchor="bottom", y=-0.22,
                xanchor="center", x=0.5),
    annotations=[dict(text="Datenquelle: OECD PISA-Studien.",
                      x=0, y=-0.32, xref="paper", yref="paper",
                      showarrow=False, font=dict(size=11, color=COLOR["neutral"]),
                      xanchor="left")],
)
fig.update_xaxes(title="Erhebungsjahr", tickmode="array", tickvals=jahre,
                 range=[anfang - 0.5, ende + 0.5])
fig.update_yaxes(title="PISA-Punktzahl", range=[y_min, y_max], dtick=10)

fig.show()
```

**Befund:** In allen drei Domänen liegen die deutschen Schülerinnen und Schüler 2022 deutlich unter dem Stand von 2009 – am stärksten in Mathematik. Die Leistungsentwicklung läuft parallel zu den zuvor beschriebenen Trends und verdichtet das Bild eines Systems unter strukturellem Druck.


**Visualisierungswahl – Multi-Linien-Diagramm:** Drei Domänen über sieben Erhebungszeitpunkte – ein Liniendiagramm mit drei farbcodierten Reihen ist die natürliche Form. Es erlaubt, sowohl den Trend jeder einzelnen Domäne als auch den parallelen Verlauf aller drei Reihen gleichzeitig zu erfassen. Die im Untertitel ausgewiesenen Δ-Werte (Anfangs- vs. Endjahr) liefern die zentralen Zahlen direkt mit, ohne dass der Leser sie aus dem Diagramm rekonstruieren muss. Der `hovermode="x unified"` zeigt beim Überfahren eines Jahres alle drei Punktwerte gemeinsam – ideal für den Domänenvergleich.

**Warum nicht anders?** Drei separate Diagramme nebeneinander hätten den parallelen Verlauf optisch zerschnitten – gerade die Gleichzeitigkeit des Rückgangs ist aber die Kernaussage. Ein Balkendiagramm pro Erhebungsjahr (gruppiert nach Domäne) wäre möglich, hätte aber den Trendverlauf durch die Säulenform schwächer transportiert. Ein normalisiertes Diagramm (Indexwert relativ zum Startjahr = 100) hätte den Rückgang noch deutlicher gezeigt, aber die absolute PISA-Skala verloren, deren Verankerung (500 Punkte Mittelwert, 20 Punkte ≈ ein Schuljahr) für die Einordnung wichtig ist.

**Nachteile, die wir in Kauf nehmen:** Drei Linien in einem Diagramm verlangen Farbunterscheidung – Leser mit Rot-Grün-Schwäche können Lesekompetenz (rot) und Naturwissenschaften (grün) nur eingeschränkt trennen. Außerdem liegen die drei Reihen auf einer gemeinsamen Y-Achse, was visuell suggeriert, dass Punktwerte zwischen den Domänen direkt vergleichbar sind – tatsächlich sind die Skalen pro Domäne unabhängig kalibriert.

**Technischer Aufbau:** Drei Excel-Dateien werden über den `load()`-Helper getrennt eingelesen und auf die Schnittmenge gemeinsamer Erhebungsjahre seit 2009 reduziert. Eine Hilfsfunktion `vals_for(df)` extrahiert pro Jahr den Punktwert, sodass alle drei Domänen synchron mit derselben Jahresachse arbeiten. Die Visualisierung nutzt *plotly* (`graph_objects`): drei `go.Scatter`-Traces im Modus `lines+markers` mit Farben aus der Projekt-Palette. Die Δ-Werte werden in einer Schleife pro Domäne berechnet und als HTML-formatierter Untertitel zusammengesetzt – inklusive farblicher Kennzeichnung. Die Y-Achsen-Grenzen entstehen dynamisch durch Auf-/Abrunden auf Zehnerschritte, damit die Skala bei Datenupdates stabil bleibt. *plotly* hier wegen `hovermode="x unified"` – diese Funktion ist für den synchronen Drei-Domänen-Vergleich beim Überfahren eines Jahres das stärkste Argument.


### 3.6 Lehrkräfte-Perspektive – Wahrgenommene Herausforderungen

Bevor wir die Befunde zusammenführen, lohnt ein Blick aus der Innensicht: Was nennen Lehrkräfte selbst als ihre größten Herausforderungen im Berufsalltag? Eine repräsentative Befragung von 1.540 Lehrkräften der Robert Bosch Stiftung (Ende 2024) liefert dazu eine geordnete Liste – und bestätigt die zuvor quantitativ gezeigten Spannungsfelder aus subjektiver Perspektive.


```python
# Wahrgenommene Herausforderungen aus Sicht der Lehrkräfte – horizontales Balkendiagramm.
#
# Datenbasis: Robert Bosch Stiftung, Befragung von 1.540 Lehrkräften
# (11.11. – 02.12.2024). Mehrfachnennungen waren möglich, daher ergeben die
# Anteile in Summe deutlich mehr als 100 Prozent.
#
# Visualisiert wird ein horizontales Balkendiagramm, in dem die Kategorien
# nach Häufigkeit sortiert sind. Die Heterogenität wird farblich
# hervorgehoben, da sie unmittelbar an die zentrale Forschungsfrage anknüpft.

herausf = (load(F_HERAUSF).rename(columns={"Schuljahr": "Herausforderung",
                                            "Anzahl":   "Anteil"})
                          .astype({"Anteil": float})
                          .sort_values("Anteil", ascending=True)
                          .reset_index(drop=True))

HIGHLIGHT = {"Heterogenität", "Personalmangel", "(zu) große Klassen", "Leistungsschwache Schüler:innen; Lernrückstände"}
farben = [COLOR["accent"] if k in HIGHLIGHT else COLOR["primary"]
          for k in herausf["Herausforderung"]]

fig, ax = plt.subplots(figsize=(10, 6.5))

bars = ax.barh(herausf["Herausforderung"], herausf["Anteil"],
               color=farben, edgecolor="white", linewidth=0.8)

for b, v in zip(bars, herausf["Anteil"]):
    ax.text(v + 0.6, b.get_y() + b.get_height() / 2,
            f"{int(v)} %", va="center", ha="left",
            fontsize=11, fontweight="bold", color=COLOR["ink"])

ax.set_xlim(0, max(herausf["Anteil"]) + 8)
ax.xaxis.set_major_formatter(mpl.ticker.PercentFormatter(decimals=0))
ax.set_xlabel("Anteil der Nennungen (Mehrfachnennungen möglich)")
ax.grid(axis="x")
ax.set_axisbelow(True)
ax.tick_params(axis="y", length=0)
ax.spines["left"].set_visible(False)

ax.set_title("Die größten Herausforderungen aus Sicht der Lehrkräfte (2024)\n"
             "Heterogenität, Personalmangel, große Klassen und Lernrückstände hervorgehoben",
             loc="left", pad=14)

fig.text(0.01, -0.02,
         "Datenquelle: Robert Bosch Stiftung, Befragung von 1.540 Lehrkräften, "
         "11.11. – 02.12.2024. Mehrfachnennungen möglich.",
         fontsize=9, color=COLOR["neutral"])

plt.tight_layout()
plt.show()
```

**Befund:**  Die Lehrkräfte 2024 nennen Verhalten der Schüler:innen (42 %) mit Abstand am häufigsten als größte Herausforderung, gefolgt von Arbeitsbelastung/Zeitmangel (34 %) und Heterogenität (32 %). Drei der Top-Belastungen — Heterogenität, Personalmangel (20 %) sowie leistungsschwache Schüler:innen/Lernrückstände (13 %) — verweisen direkt auf strukturelle Engpässe: Klassen werden vielfältiger, gleichzeitig fehlt Personal, um dieser Vielfalt gerecht zu werden. Die Befragung von 1.540 Lehrkräften zeichnet damit das Bild eines Systems, das unter dem Gewicht heterogener Lerngruppen bei knapper Personaldecke arbeitet.


**Visualisierungswahl – horizontales Balkendiagramm:** Zehn Kategorien mit langen deutschen Labels („Leistungsschwache Schüler:innen; Lernrückstände", „Sanierung, Renovierung, Investition") lassen sich horizontal lesbar darstellen, ohne dass Beschriftungen gedreht oder abgeschnitten werden müssen. Die Sortierung nach Häufigkeit absteigend führt das Auge automatisch von der größten zur kleinsten Nennung — Ranking-Information ist auf einen Blick erfassbar. Die farbliche Hervorhebung (rot) der drei kapazitätsbezogenen Items (Heterogenität, Personalmangel, große Klassen) plus Lernrückstände trennt visuell die strukturellen Belastungen von den situativen (Verhalten, Eltern, Bürokratie) und macht den eigentlichen Befund — ein Kapazitätsproblem — ohne zusätzlichen Erklärungstext sichtbar.

**Warum nicht anders?** Ein Tortendiagramm scheidet aus, weil Mehrfachnennungen möglich waren — die Anteile summieren sich nicht auf 100 % und eine Kreisaufteilung wäre schlicht falsch. Ein vertikales Säulendiagramm hätte die langen Kategorienamen entweder gedreht (schlecht lesbar) oder umgebrochen (unruhiges Layout) dargestellt. Eine Wortwolke hätte zwar die Häufigkeitsdimension transportiert, aber keine präzisen Prozentwerte und keine sinnvolle Rangordnung erlaubt. Ein Treemap wäre für zehn Items denkbar, hätte aber den linearen Vergleich „wie viel mehr ist X gegenüber Y" durch Flächenvergleich ersetzt — kognitiv aufwendiger als Balkenlängen.

**Nachteile, die wir in Kauf nehmen:** Die Zweifarbigkeit (rot/blau) ist eine interpretative Setzung der Visualisierung — wir gruppieren kapazitätsbezogen vs. übrige Belastungen, was die Originaldaten so nicht vorgeben. Leser:innen mit Rot-Grün-Schwäche können die Hervorhebung schwerer erkennen; die Reihenfolge bleibt aber auch ohne Farbe lesbar. Außerdem zeigt das Diagramm nur die Lehrersicht — Eltern oder Schüler:innen würden andere Ranglisten produzieren, und die einzelnen Items sind unterschiedlich operationalisierbar („Verhalten" ist breiter gefasst als „Digitalisierung allgemein"). Schließlich gehen Verteilungsinformationen innerhalb der Lehrerschaft (Schulform, Region, Berufserfahrung) durch die Aggregation verloren.

**Technischer Aufbau:** Die Daten liegen als DataFrame mit zwei Spalten (Kategorie, Anteil) vor und werden vor dem Plotten nach Anteil aufsteigend sortiert — barh() zeichnet von unten nach oben, sodass aufsteigende Sortierung visuell als absteigende Rangliste erscheint. Eine Boolean-Maske (highlight = kategorien.isin([...])) erzeugt die Farbliste über einen List-Comprehension-Ausdruck mit den Projektfarben #C44E2C (rot) und #2C3E5E (blau). Die Prozentwerte werden per Schleife mit ax.text() rechts neben jeden Balken gesetzt, xaxis.set_major_formatter(PercentFormatter(xmax=100)) liefert die Achsenbeschriftung. Die Quelle steht als figtext() unten links im grauen Kleindruck, der Titel zweizeilig über \n mit Untertitel zur Befundzusammenfassung. tight_layout() plus figsize=(10, 6) verhindern abgeschnittene Labels bei den langen Kategorienamen.


## 4. Fazit

Die Analyse zeigt, dass die scheinbar positive Entwicklung der Schüler-Lehrer-Relation bei genauerem Hinsehen ein verzerrtes Bild liefert. Drei Befunde sind zentral:

1. **Lehrkapazität ist niedriger als die SLR suggeriert.** Rund jede zweite Lehrkraft arbeitet nicht in Vollzeit. Rechnet man die Teilzeit- und Stundenweise-Anteile in Vollzeitäquivalente um, schrumpft die reale Lehrkapazität deutlich gegenüber der Kopfzahl.
2. **Die Anforderungen wachsen schneller als die Ressourcen.** Die Zahl der Schülerinnen und Schüler mit sonderpädagogischem Förderbedarf an Regelschulen hat sich von rund 157.000 (2013) auf rund 264.000 (2023) fast verdoppelt – während die Förderschulzahlen nahezu konstant blieben. Die KMK-Prognose erwartet bis Anfang der 2030er Jahre einen anhaltenden Mangel an Lehrkräften.
3. **Die Folgen sind messbar.** Die PISA-Ergebnisse zeigen seit 2012 einen klaren Leistungsrückgang in allen drei Kernkompetenzen. Auch die Lehrkräfte selbst nennen Heterogenität, Personalmangel und Lernrückstände als zentrale Herausforderungen ihres Berufsalltags.

**Antwort auf die Forschungsfrage:** Über die letzten 15 Jahre ist die reale Lehrkapazität – gemessen in Vollzeitäquivalenten und effektiver Verfügbarkeit – nicht mit der wachsenden Heterogenität in deutschen Schulklassen mitgewachsen. Individuelle Förderung gemäß dem KMK-Anspruch ist unter diesen Rahmenbedingungen für viele Lehrkräfte strukturell nicht leistbar – die Vision der „Bildungsrepublik“ trifft im Alltag auf systemische Grenzen.
